```python
class Solution:
    def sort012(self, arr):

        count0 = 0
        count1 = 0
        count2 = 0

        for num in arr:
            if num == 0:
                count0 += 1
            elif num == 1:
                count1 += 1
            else:
                count2 += 1

        index = 0

        for i in range(count0):
            arr[index] = 0
            index += 1

        for i in range(count1):
            arr[index] = 1
            index += 1

        for i in range(count2):
            arr[index] = 2
            index += 1
```

```python
class Solution:
    def firstSearch(self, arr, k):
        left = 0
        right = len(arr) - 1
        result = -1

        while left <= right:
            mid = (left + right) // 2

            if arr[mid] == k:
                result = mid
                right = mid - 1
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1

        return result
```
