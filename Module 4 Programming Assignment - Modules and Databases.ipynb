```python
import sqlite3

conn = sqlite3.connect("books.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE books (
    title TEXT,
    author TEXT,
    year INTEGER
)
""")

conn.commit()
conn.close()
```

```python
import csv
import sqlite3

conn = sqlite3.connect("books.db")
cursor = conn.cursor()

with open("books2.csv", newline="") as csvfile:
    reader = csv.reader(csvfile)
    next(reader)          # Skip the header row

    for row in reader:
        cursor.execute(
            "INSERT INTO books VALUES (?, ?, ?)",
            row
        )

conn.commit()
conn.close()
```
